```python
# Célula 1 — Imports
import pandas as pd
from pathlib import Path
```

```python
# Célula 2 — Listar arquivos na pasta SIAT
pasta = Path(r"C:\Users\franc\OneDrive\Francisco\Profissional\MBA_Data_Science_ e_Analytics\00_TCC\06_Dados_base\GEO\2023_11_03_SIAT")

arquivos = sorted(pasta.rglob("*"))
print(f"Arquivos encontrados: {len(arquivos)}\n")
for a in arquivos:
    tamanho_mb = a.stat().st_size / 1024 / 1024
    print(f"  {a.name:<60} {tamanho_mb:>8.1f} MB")
```

```python
# Célula 3 — Inspecionar primeiras linhas de cada .txt
txts = sorted(pasta.glob("*.txt"))
print(f"\n\nArquivos .txt: {len(txts)}")

for txt in txts:
    print(f"\n{'='*60}")
    print(f"{txt.name}")
    print("="*60)
    # Tentar ler com separadores comuns
    for sep in ["|", ";", "\t", ","]:
        try:
            df = pd.read_csv(txt, sep=sep, nrows=2, encoding="latin1")
            if len(df.columns) > 1:
                print(f"Separador: '{sep}' | Colunas: {len(df.columns)}")
                print(f"Colunas: {df.columns.tolist()}")
                print(f"Amostra:")
                print(df.head(2).to_string())
                break
        except Exception as e:
            continue
```
