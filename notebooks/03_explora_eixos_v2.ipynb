```python
# Célula 1 — Imports
import geopandas as gpd
import pandas as pd
from pathlib import Path
```

```python
# Célula 2 — Carregar eixos
pasta = Path(r"C:\Users\franc\OneDrive\Francisco\Profissional\MBA_Data_Science_ e_Analytics\00_TCC\06_Dados_base\GEO\eixos")
eixos = gpd.read_file(pasta / "eixos.shp")
```

```python
# Célula 3 — Categorias de vias
print("Categorias de vias (CDIDECAT):")
print(eixos["CDIDECAT"].value_counts().to_string())
```

```python
# Célula 4 — Comprimento total por categoria
eixos["COMP_M"] = eixos.geometry.length.round(1)
print("\nComprimento total por categoria (metros):")
resumo = eixos.groupby("CDIDECAT")["COMP_M"].agg(
    qtd_trechos="count",
    comp_total_m="sum",
    comp_medio_m="mean"
).sort_values("comp_total_m", ascending=False)
resumo["comp_total_km"] = (resumo["comp_total_m"] / 1000).round(1)
resumo["comp_medio_m"]  = resumo["comp_medio_m"].round(1)
print(resumo.to_string())
```

```python
# Célula 5 — Amostra de avenidas
print("\nAmostra de Avenidas (CDIDECAT = 'AV'):")
avs = eixos[eixos["CDIDECAT"] == "AV"][["NMIDELOG", "NMIDEABR", "COMP_M"]].copy()
avs_agrup = avs.groupby("NMIDELOG")["COMP_M"].sum().sort_values(ascending=False).head(20)
print(avs_agrup.round(0).to_string())
```
