```python
# Célula 1 — Imports
import pandas as pd
from pathlib import Path
```

```python
# Célula 2 — Carregar todas as abas do arquivo
arquivo = Path(r"C:\Users\franc\OneDrive\Francisco\Profissional\MBA_Data_Science_ e_Analytics\00_TCC\06_Dados_base\GEO\IMO_GRD.xlsx")

xl = pd.ExcelFile(arquivo)
print(f"Abas encontradas ({len(xl.sheet_names)}):")
for aba in xl.sheet_names:
    print(f"  {aba}")
```

```python
# Célula 3 — Inspecionar cada aba
abas_desconsiderar = ["IMO_CLASSES_E_RH"]

for aba in xl.sheet_names:
    df = pd.read_excel(arquivo, sheet_name=aba, nrows=3)
    print(f"\n{'='*60}")
    print(f"ABA: {aba}")
    if aba in abas_desconsiderar:
        print("  [DESCONSIDERADA — RH da época, uso futuro apenas para comparação]")
    print(f"{'='*60}")
    print(f"Colunas ({len(df.columns)}):")
    df_full = pd.read_excel(arquivo, sheet_name=aba)
    for col in df_full.columns:
        nulos = df_full[col].isna().sum()
        pct   = nulos / len(df_full) * 100
        dtype = str(df_full[col].dtype)
        print(f"  {col:<40} {dtype:<15} nulos: {nulos:,} ({pct:.1f}%)")
    print(f"\nTotal de registros: {len(df_full):,}")
    print(f"\nAmostra (3 registros):")
    print(df_full.head(3).to_string())
```
