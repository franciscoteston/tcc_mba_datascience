```python
# Célula 1 — Imports
import geopandas as gpd
import pandas as pd
from pathlib import Path
```

```python
# Célula 2 — Definir pastas
pastas = {
    "limite_poa":    Path(r"C:\Users\franc\OneDrive\Francisco\Profissional\MBA_Data_Science_ e_Analytics\00_TCC\06_Dados_base\GEO\limite_poa"),
    "poa_ilhas":     Path(r"C:\Users\franc\OneDrive\Francisco\Profissional\MBA_Data_Science_ e_Analytics\00_TCC\06_Dados_base\GEO\poa_ilhas"),
    "bairros_vigentes": Path(r"C:\Users\franc\OneDrive\Francisco\Profissional\MBA_Data_Science_ e_Analytics\00_TCC\06_Dados_base\GEO\bairros_vigentes"),
}
```

```python
# Célula 3 — Inspecionar cada pasta e shapefile
for nome, pasta in pastas.items():
    shapefiles = list(pasta.rglob("*.shp"))
    print(f"\n{'='*60}")
    print(f"PASTA: {nome} ({len(shapefiles)} shapefile(s))")
    print("="*60)
    for shp in sorted(shapefiles):
        gdf = gpd.read_file(shp)
        print(f"\nArquivo: {shp.name}")
        print(f"Registros:  {len(gdf):,}")
        print(f"CRS:        {gdf.crs}")
        print(f"Geometria:  {gdf.geom_type.unique()}")
        print(f"Colunas ({len(gdf.columns)}):")
        for col in gdf.columns:
            nulos = gdf[col].isna().sum()
            pct   = nulos / len(gdf) * 100
            print(f"  {col:<30} {str(gdf[col].dtype):<15} nulos: {nulos:,} ({pct:.1f}%)")
        print(f"\nAmostra (3 registros):")
        print(gdf.drop(columns="geometry").head(3).to_string())
```
